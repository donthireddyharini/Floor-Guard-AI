```python
# Step 3: Run YOLOv8s Training
from ultralytics import YOLO

# Load pretrained YOLOv8s model
model = YOLO('yolov8s.pt')

# Train on FloorGuard floor hazard dataset
results = model.train(
    data='floorguard_yolo.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    project='floorguard_runs',
    name='yolov8s_floorguard'
)
```

```python
# Step 4: Evaluate Model on Held-out Test Set
metrics = model.val(split='test')
print(f"mAP50: {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
```

```python
# Step 5: Export best checkpoint
import shutil
shutil.copy('floorguard_runs/yolov8s_floorguard/weights/best.pt', 'floorguard_yolov8s_best.pt')
print("Checkpoint ready for export: floorguard_yolov8s_best.pt")
```
